```python
import pandas as pd
import numpy as np
from scipy.sparse import csr_matrix
from sklearn.neighbors import NearestNeighbors
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from math import sqrt
```

```python
# ==========================================
# 0. Глобальные параметры
# ==========================================
K_NEIGHBORS_USERS = 50      
K_NEIGHBORS_ITEMS = 20      
TOP_X_RECS = 10             
TEST_SIZE = 0.2
SAMPLE_USER_ID = 2

ALPHA = 0.6                 # Вес User-Based
INTERSECTION_BONUS = 0.3    # Бонус за пересечение UB и IB
SIMILARITY_THRESHOLD = 0.2  # Порог отсечения шума

def load_movielens_data():
    ratings = pd.read_csv('ratings.dat', sep='::', engine='python',
                          names=['userId', 'movieId', 'rating', 'timestamp'], encoding='ISO-8859-1')
    movies = pd.read_csv('movies.dat', sep='::', engine='python',
                         names=['movieId', 'title', 'genres'], encoding='ISO-8859-1')
    users = pd.read_csv('users.dat', sep='::', engine='python',
                        names=['userId', 'gender', 'age', 'occupation', 'zip-code'], encoding='ISO-8859-1')
    return ratings, movies, users

def create_matrices(df):
    # Матрица Пользователь x Фильм (для User-Based)
    pivot_users = df.pivot(index='userId', columns='movieId', values='rating').fillna(0)
    sparse_users = csr_matrix(pivot_users.values)
    
    # Матрица Фильм x Пользователь (для Item-Based)
    pivot_items = df.pivot(index='movieId', columns='userId', values='rating').fillna(0)
    sparse_items = csr_matrix(pivot_items.values)
    
    return pivot_users, sparse_users, pivot_items, sparse_items
```

```python
# ==========================================
# 1. User-Based предсказания
# ==========================================
def get_ub_predictions(user_id, train_pivot_users, model_knn_users, k_neighbors):
    if user_id not in train_pivot_users.index:
        return {}
    
    user_idx = train_pivot_users.index.get_loc(user_id)
    user_vector = train_pivot_users.iloc[user_idx, :].values.reshape(1, -1)
    
    distances, indices = model_knn_users.kneighbors(user_vector, n_neighbors=k_neighbors + 1)
    
    ub_scores = {}
    user_watched = train_pivot_users.iloc[user_idx, :][train_pivot_users.iloc[user_idx, :] > 0].index.tolist()
    user_avg = train_pivot_users.iloc[user_idx, :][train_pivot_users.iloc[user_idx, :] > 0].mean() if len(user_watched) > 0 else 3.0
    
    for i in range(1, len(distances.flatten())):
        similarity = 1 - distances.flatten()[i]
        if similarity < SIMILARITY_THRESHOLD:
            continue
            
        neighbor_idx = indices.flatten()[i]
        neighbor_ratings_vec = train_pivot_users.iloc[neighbor_idx, :]
        nonzero_ratings = neighbor_ratings_vec[neighbor_ratings_vec > 0]
        
        for movie_id, rating in nonzero_ratings.items():
            if movie_id not in user_watched:
                if movie_id not in ub_scores:
                    ub_scores[movie_id] = {'weighted_sum': 0, 'weight_sum': 0}
                ub_scores[movie_id]['weighted_sum'] += rating * similarity
                ub_scores[movie_id]['weight_sum'] += similarity
                
    final_ub = {}
    for movie_id, scores in ub_scores.items():
        if scores['weight_sum'] > 0:
            raw_pred = scores['weighted_sum'] / scores['weight_sum']
            norm_pred = raw_pred * 0.8 + user_avg * 0.2
            final_ub[movie_id] = max(1.0, min(5.0, norm_pred))
            
    return final_ub
```

```python
# ==========================================
# 2. Item-Based предсказания (ОПТИМИЗИРОВАННАЯ ВЕРСИЯ)
# ==========================================
def get_ib_predictions(user_id, train_pivot_users, train_pivot_items, model_knn_items, k_neighbors):
    if user_id not in train_pivot_users.index:
        return {}
        
    user_ratings = train_pivot_users.loc[user_id]
    # Оставляем только те фильмы, которые физически есть в матрице items
    watched_movies = [mid for mid in user_ratings[user_ratings > 0].index if mid in train_pivot_items.index]
    
    if not watched_movies:
        return {}
        
    # Получаем индексы и векторы для ВСЕХ просмотренных фильмов СРАЗУ
    m_indices = [train_pivot_items.index.get_loc(mid) for mid in watched_movies]
    m_vectors = train_pivot_items.iloc[m_indices, :].values
    
    # Вызываем kneighbors ОДИН РАЗ для всего батча (это и есть главное ускорение)
    distances, indices = model_knn_items.kneighbors(m_vectors, n_neighbors=k_neighbors + 1)
    
    ib_scores = {}
    watched_set = set(watched_movies)
    
    # Обрабатываем результаты батча
    for i, movie_id in enumerate(watched_movies):
        user_rating = user_ratings[movie_id]
        
        # distances[i] и indices[i] соответствуют movie_id
        for j in range(1, len(distances[i])):
            similarity = 1 - distances[i][j]
            if similarity < SIMILARITY_THRESHOLD:
                continue
                
            neighbor_idx = indices[i][j]
            neighbor_movie_id = train_pivot_items.index[neighbor_idx]
            
            if neighbor_movie_id not in watched_set:
                if neighbor_movie_id not in ib_scores:
                    ib_scores[neighbor_movie_id] = {'weighted_sum': 0.0, 'weight_sum': 0.0}
                
                ib_scores[neighbor_movie_id]['weighted_sum'] += user_rating * similarity
                ib_scores[neighbor_movie_id]['weight_sum'] += similarity
                
    final_ib = {}
    for movie_id, scores in ib_scores.items():
        if scores['weight_sum'] > 0:
            final_ib[movie_id] = max(1.0, min(5.0, scores['weighted_sum'] / scores['weight_sum']))
            
    return final_ib
```

```python
# ==========================================
# 3. Гибридное объединение
# ==========================================
def get_hybrid_recommendations(user_id, train_pivot_users, train_pivot_items, 
                               model_knn_users, model_knn_items, movie_titles, top_n=10, alpha=0.6):
    
    ub_preds = get_ub_predictions(user_id, train_pivot_users, model_knn_users, K_NEIGHBORS_USERS)
    ib_preds = get_ib_predictions(user_id, train_pivot_users, train_pivot_items, model_knn_items, K_NEIGHBORS_ITEMS)
    
    all_recommended_movies = set(ub_preds.keys()).union(set(ib_preds.keys()))
    
    hybrid_scores = []
    
    for movie_id in all_recommended_movies:
        score_ub = ub_preds.get(movie_id, 0.0)
        score_ib = ib_preds.get(movie_id, 0.0)
        
        in_both = 1.0 if (score_ub > 0 and score_ib > 0) else 0.0
        
        if score_ub > 0 and score_ib == 0:
            final_score = score_ub
        elif score_ib > 0 and score_ub == 0:
            final_score = score_ib
        else:
            final_score = (alpha * score_ub) + ((1 - alpha) * score_ib) + (INTERSECTION_BONUS * in_both)
            
        final_score = max(1.0, min(5.0, final_score))
        
        hybrid_scores.append({
            'movieId': movie_id,
            'title': movie_titles.get(movie_id, "Unknown"),
            'predicted_rating': final_score,
            'ub_score': score_ub,
            'ib_score': score_ib,
            'in_both': bool(in_both)
        })
        
    hybrid_scores.sort(key=lambda x: x['predicted_rating'], reverse=True)
    return hybrid_scores[:top_n]
```

```python
# ==========================================
# 4. Оценка качества (Метрики)
# ==========================================
def evaluate_hybrid_model(test_df, train_pivot_users, train_pivot_items, model_knn_users, model_knn_items, movie_titles, k_users, k_items, top_k=10):
    actual_ratings = []
    predicted_ratings = []
    
    precisions = []
    recalls = []
    ndcgs = []
    
    test_users = test_df['userId'].unique()
    sample_users = np.random.choice(test_users, min(200, len(test_users)), replace=False)
    
    for user_id in sample_users:
        user_test_data = test_df[test_df['userId'] == user_id]
        relevant_movies = set(user_test_data[user_test_data['rating'] >= 4.0]['movieId'].tolist())
        
        if not relevant_movies:
            continue
            
        # Вызов с именованными аргументами для надежности
        recs = get_hybrid_recommendations(
            user_id=user_id, 
            train_pivot_users=train_pivot_users, 
            train_pivot_items=train_pivot_items,
            model_knn_users=model_knn_users, 
            model_knn_items=model_knn_items, 
            movie_titles=movie_titles, 
            top_n=top_k
        )
        
        recommended_movies = [r['movieId'] for r in recs]
        
        for r in recs:
            mid = r['movieId']
            actual = user_test_data[user_test_data['movieId'] == mid]['rating']
            if not actual.empty:
                predicted_ratings.append(r['predicted_rating'])
                actual_ratings.append(actual.values[0])
                
        hits = len(set(recommended_movies).intersection(relevant_movies))
        
        precisions.append(hits / top_k if top_k > 0 else 0)
        recalls.append(hits / len(relevant_movies) if len(relevant_movies) > 0 else 0)
        
        dcg = 0.0
        for i, mid in enumerate(recommended_movies):
            if mid in relevant_movies:
                rel = user_test_data[user_test_data['movieId'] == mid]['rating'].values[0]
                dcg += (rel) / np.log2(i + 2)
                
        ideal_rels = sorted(user_test_data[user_test_data['movieId'].isin(relevant_movies)]['rating'].tolist(), reverse=True)[:top_k]
        idcg = sum([(rel) / np.log2(i + 2) for i, rel in enumerate(ideal_rels)])
        
        ndcgs.append(dcg / idcg if idcg > 0 else 0)

    rmse = sqrt(mean_squared_error(actual_ratings, predicted_ratings)) if actual_ratings else 0
    
    return {
        'RMSE': rmse,
        'Precision@K': np.mean(precisions),
        'Recall@K': np.mean(recalls),
        'NDCG@K': np.mean(ndcgs)
    }
```

```python
# ==========================================
# 5. Основная программа
# ==========================================
if __name__ == "__main__":
    print("Загрузка данных...")
    ratings_df, movies_df, users_df = load_movielens_data()
    movie_titles = dict(zip(movies_df['movieId'], movies_df['title']))
    
    print("Разделение на train/test...")
    train_data, test_data = train_test_split(ratings_df, test_size=TEST_SIZE, random_state=42)
    
    print("Создание матриц...")
    train_pivot_users, train_sparse_users, train_pivot_items, train_sparse_items = create_matrices(train_data)
    
    print("Обучение User-Based модели...")
    model_knn_users = NearestNeighbors(metric='cosine', algorithm='brute', n_neighbors=K_NEIGHBORS_USERS, n_jobs=-1)
    model_knn_users.fit(train_sparse_users)
    
    print("Обучение Item-Based модели...")
    model_knn_items = NearestNeighbors(metric='cosine', algorithm='brute', n_neighbors=K_NEIGHBORS_ITEMS, n_jobs=-1)
    model_knn_items.fit(train_sparse_items)
    
    print(f"\nГенерация рекомендаций для пользователя ID={SAMPLE_USER_ID}...")
    
    # ЯВНЫЙ ВЫЗОВ С ИМЕНОВАННЫМИ АРГУМЕНТАМИ (гарантирует отсутствие ошибок позиционирования)
    hybrid_recs = get_hybrid_recommendations(
        user_id=SAMPLE_USER_ID, 
        train_pivot_users=train_pivot_users, 
        train_pivot_items=train_pivot_items, 
        model_knn_users=model_knn_users, 
        model_knn_items=model_knn_items, 
        movie_titles=movie_titles, 
        top_n=10
    )
    
    print("\nТоп-10 гибридных рекомендаций:")
    for i, rec in enumerate(hybrid_recs, 1):
        in_both_str = " [UB+IB]" if rec['in_both'] else ""
        print(f"{i:2}. {rec['title']:<45} (Оценка: {rec['predicted_rating']:.2f}, UB: {rec['ub_score']:.2f}, IB: {rec['ib_score']:.2f}){in_both_str}")
        
    print("\n" + "="*60)
    print("ОЦЕНКА КАЧЕСТВА МОДЕЛИ (на подвыборке 200 пользователей)")
    print("="*60)
    
    metrics = evaluate_hybrid_model(
        test_df=test_data, 
        train_pivot_users=train_pivot_users, 
        train_pivot_items=train_pivot_items, 
        model_knn_users=model_knn_users, 
        model_knn_items=model_knn_items, 
        movie_titles=movie_titles, 
        k_users=K_NEIGHBORS_USERS, 
        k_items=K_NEIGHBORS_ITEMS, 
        top_k=10
    )
    
    print(f"RMSE:          {metrics['RMSE']:.4f}")
    print(f"Precision@10:  {metrics['Precision@K']:.4f}")
    print(f"Recall@10:     {metrics['Recall@K']:.4f}")
    print(f"NDCG@10:       {metrics['NDCG@K']:.4f}")
```

## тут код для отчетника и тд 

```python
print("="*60)
print("ИССЛЕДОВАНИЕ 1: Активные vs Неактивные пользователи")
print("="*60)

# 1. Считаем количество оценок каждого пользователя в трейне
user_activity = train_data.groupby('userId').size()

# 2. Находим пороги для 10% самых активных и 10% самых неактивных
threshold_high = user_activity.quantile(0.90) # Топ 10%
threshold_low = user_activity.quantile(0.10)  # Низ 10%

print(f"Порог 'активного' пользователя (90-й перцентиль): > {int(threshold_high)} оценок")
print(f"Порог 'неактивного' пользователя (10-й перцентиль): <= {int(threshold_low)} оценок")

# 3. Функция для расчета метрик на конкретном списке пользователей
def evaluate_specific_users(user_list, test_df, desc):
    if len(user_list) == 0:
        return {'Precision@10': 0, 'Recall@10': 0, 'NDCG@10': 0, 'Count': 0}
    
    precisions, recalls, ndcgs = [], [], []
    top_k = 10
    
    # Уменьшили выборку до 50 для ускорения (статистической значимости хватит)
    sample_users = np.random.choice(user_list, min(50, len(user_list)), replace=False)
    
    for user_id in sample_users:
        user_test_data = test_df[test_df['userId'] == user_id]
        relevant_movies = set(user_test_data[user_test_data['rating'] >= 4.0]['movieId'].tolist())
        if not relevant_movies: continue
            
        recs = get_hybrid_recommendations(
            user_id=user_id, train_pivot_users=train_pivot_users, train_pivot_items=train_pivot_items,
            model_knn_users=model_knn_users, model_knn_items=model_knn_items, 
            movie_titles=movie_titles, top_n=top_k
        )
        recommended_movies = [r['movieId'] for r in recs]
        hits = len(set(recommended_movies).intersection(relevant_movies))
        
        precisions.append(hits / top_k)
        recalls.append(hits / len(relevant_movies))
        
        # Упрощенный NDCG
        dcg = sum([1 / np.log2(i + 2) for i, mid in enumerate(recommended_movies) if mid in relevant_movies])
        idcg = sum([1 / np.log2(i + 2) for i in range(min(len(relevant_movies), top_k))])
        ndcgs.append(dcg / idcg if idcg > 0 else 0)

    return {
        'Precision@10': np.mean(precisions) if precisions else 0,
        'Recall@10': np.mean(recalls) if recalls else 0,
        'NDCG@10': np.mean(ndcgs) if ndcgs else 0,
        'Count': len(sample_users)
    }

# 4. Запуск сравнения
active_users = user_activity[user_activity >= threshold_high].index.tolist()
inactive_users = user_activity[user_activity <= threshold_low].index.tolist()

# Фильтруем только тех, кто есть в тесте
active_users_test = list(set(active_users).intersection(test_data['userId'].unique()))
inactive_users_test = list(set(inactive_users).intersection(test_data['userId'].unique()))

metrics_active = evaluate_specific_users(active_users_test, test_data, "Active")
metrics_inactive = evaluate_specific_users(inactive_users_test, test_data, "Inactive")

print(f"\n--- Метрики для ТОП-10% АКТИВНЫХ (оценок >= {int(threshold_high)}) ---")
print(f"Precision@10: {metrics_active['Precision@10']:.4f} | Recall@10: {metrics_active['Recall@10']:.4f} | NDCG@10: {metrics_active['NDCG@10']:.4f}")

print(f"\n--- Метрики для НИЗ-10% НЕАКТИВНЫХ (оценок <= {int(threshold_low)}) ---")
print(f"Precision@10: {metrics_inactive['Precision@10']:.4f} | Recall@10: {metrics_inactive['Recall@10']:.4f} | NDCG@10: {metrics_inactive['NDCG@10']:.4f}")
```

## Исследование 1: Влияние активности пользователей на качество рекомендаций
Требовалось оценить, как количество исторических данных (оценок) пользователя влияет на точность работы гибридной системы (UB + IB).
Методика: Все пользователи были разделены на две группы:
Активные (топ-10% по количеству оценок в обучающей выборке).
Неактивные (низ-10% по количеству оценок).

Для обеих групп были рассчитаны метрики Precision@10, Recall@10 и NDCG@10 на тестовой выборке.

Результаты и Выводы:

У активных пользователей метрики значительно выше. Это объясняется тем, что для User-Based подхода проще найти релевантных "соседей"
так как вектор оценок плотный), а для Item-Based у пользователя больше просмотренных фильмов, от которых можно отталкиваться для поиска похожих.
У неактивных пользователей метрики (особенно Recall) падают. Системе не хватает "зацепок": мало оценок -> мало похожих пользователей и мало триггерных фильмов для IB-алгоритма.
Для улучшения работы на неактивных пользователях необходимо увеличивать вес Item-Based подхода (так как даже 1-2 оцененных нишевых фильма могут дать точных соседей-айтемов) или привлекать контентные признаки (жанры)

```python
print("="*60)
print("ИССЛЕДОВАНИЕ 2: Имитация Холодного Старта (Cold Start)")
print("="*60)

# 1. Создаем глобальный fallback-список (Топ фильмов по средней оценке, мин. 500 оценок)
movie_stats = train_data.groupby('movieId').agg({'rating': ['mean', 'count']}).reset_index()
movie_stats.columns = ['movieId', 'avg_rating', 'count']
# Фильтруем только популярные фильмы, чтобы не рекомендовать нишевый мусор
popular_movies = movie_stats[movie_stats['count'] >= 500].sort_values('avg_rating', ascending=False)
fallback_recs = popular_movies['movieId'].head(50).tolist() # Топ-50 глобальных фильмов

print(f"Создан список Fallback из {len(fallback_recs)} популярных фильмов.")

# 2. Имитируем холодный старт: берем случайных пользователей из теста
test_users = test_data['userId'].unique()
# Берем 100 случайных пользователей для оценки
sample_cold_users = np.random.choice(test_users, min(100, len(test_users)), replace=False)

print(f"Имитируем холодный старт для {len(sample_cold_users)} случайных пользователей.")
print("(Мы 'скрываем' их историю от модели и выдаем ТОЛЬКО глобальный Топ, исключая просмотренное)")

precisions_cold, recalls_cold, ndcgs_cold = [], [], []
top_k = 10

for user_id in sample_cold_users:
    user_test_data = test_data[test_data['userId'] == user_id]
    
    # Для холодного старта порог 3.0 более реалистичен, так как мы не знаем точных вкусов
    relevant_movies = set(user_test_data[user_test_data['rating'] >= 3.0]['movieId'].tolist())
    
    if not relevant_movies:
        continue
        
    # Получаем фильмы, которые пользователь УЖЕ смотрел (чтобы не рекомендовать их снова)
    user_train_watched = set(train_data[train_data['userId'] == user_id]['movieId'].tolist())
    
    # Формируем выдачу ИСКЛЮЧИТЕЛЬНО из fallback, пропуская просмотренные
    recommended_movies = []
    for mid in fallback_recs:
        if mid not in user_train_watched:
            recommended_movies.append(mid)
        if len(recommended_movies) >= top_k:
            break
            
    # Считаем метрики
    hits = len(set(recommended_movies).intersection(relevant_movies))
    
    precisions_cold.append(hits / top_k)
    recalls_cold.append(hits / len(relevant_movies))
    
    # Упрощенный NDCG (бинарная релевантность: попал в список или нет)
    dcg = sum([1 / np.log2(i + 2) for i, mid in enumerate(recommended_movies) if mid in relevant_movies])
    idcg = sum([1 / np.log2(i + 2) for i in range(min(len(relevant_movies), top_k))])
    ndcgs_cold.append(dcg / idcg if idcg > 0 else 0)

print("\n--- Метрики для ИМИТИРУЕМЫХ 'холодных' пользователей (только Popularity Fallback) ---")
if len(precisions_cold) > 0:
    print(f"Успешно рассчитано для {len(precisions_cold)} пользователей (порог релевантности >= 3.0).")
    print(f"Precision@10: {np.mean(precisions_cold):.4f}")
    print(f"Recall@10:    {np.mean(recalls_cold):.4f}")
    print(f"NDCG@10:      {np.mean(ndcgs_cold):.4f}")
else:
    print("Не удалось рассчитать метрики (нет релевантных фильмов в тесте).")
```

Вывод: При полном отсутствии истории система деградирует до рекомендации 'популярного'.
Это гарантирует ненулевой Precision, но Recall и NDCG будут ниже, чем у персонализированных
рекомендаций, так как мы не учитываем индивидуальные вкусы пользователя.

## Исследование 2: Решение проблемы "Холодного старта" (Cold Start)
Требовало разработать и оценить механику выдачи рекомендаций для совершенно новых пользователей, у которых в системе нет истории оценок.

Классическая проблема холодного старта возникает, когда у нового пользователя отсутствует история взаимодействий, из-за чего алгоритмы коллаборативной фильтрации (User-Based и Item-Based) не могут найти релевантных соседей или триггерные айтемы.

Для выполнения задания проблема была смоделирована искусственно. Для случайной выборки пользователей из тестового набора их персонализированная выдача (гибрид UB + IB) была принудительно отключена. Вместо нее система использовала механизм Popularity Fallback (резервная выдача по популярности): пользователю рекомендовались топ-N самых популярных и высокооцененных фильмов из обучающей выборки (с фильтром минимального количества голосов ≥500 для отсеивания нишевого «шума»), за исключением тех фильмов, которые пользователь уже просмотрел.

Результаты:

Анализ показал прямую зависимость точности гибридной модели от объема исторических данных пользователя:
Активные пользователи демонстрируют значительно более высокие значения Precision, Recall и NDCG. Плотный вектор оценок позволяет алгоритмам (как User-Based, так и Item-Based) надежно находить релевантных соседей и точные триггерные фильмы.
Неактивные пользователи сталкиваются с проблемой разреженности данных. Недостаток оценок приводит к снижению метрик (особенно Recall), так как системе не хватает «зацепок» для формирования персонализированного ранжированного списка.

Пути улучшения для реальных систем:
Внедрение этапа онбординга (просьба оценить 3–5 фильмов разных жанров при регистрации).
Использование демографического фолбэка (рекомендация популярного контента в возрастной/гендерной группе пользователя).
Переключение на контентные признаки (жанры, режиссеры) при первом же неявном взаимодействии (клик, добавление в «буду смотреть»).
